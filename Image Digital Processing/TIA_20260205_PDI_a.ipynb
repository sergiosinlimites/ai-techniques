<h2 style="font-size:200%; margin-top:25px;">Imágenes bitmap</h2>

<ul style="font-size:170%; line-height:35px; margin-left:50px; margin-right:50px;">
  <li>Es una <span style="color:blue">función \(f(x,y)\)</span> donde \(x\) y \(y\) son las coordenas espaciales, y el valor de \(f\) en cualquier punto \((x,y)\) es proporcional al brillo o al valor de gris de la imagen en ese punto.</li>
  <li>Pueden pensarse como construidas mediante una gran cantidad de cuadraditos, llamados <span style="color:blue">píxeles</span>, cada uno de ellos rellenos de un color uniforme.</li>
  <li>La calidad de la imagen dependerá de la cantidad de píxeles utilizados para representarla.</li>
  <li>Es el tipo de imagen utilizado para la fotografía y el cine.</li>
</ul>

<h2 style="font-size:200%; text-align:center; color:#00008B; margin-top:20px; margin-bottom:10px;">
  Digitalización de una señal analógica
</h2>

<ul style="font-size:170%; line-height:35px; margin-left:10%; margin-right:10%;">
  <li>Una señal continua o analógica posee valores definidos en cada instante de tiempo.</li>
  <li>En contraste, una señal discreta o digital sólo toma valores en instantes específicos, separados a intervalos regulares.</li>
  <li>Por tanto, una señal discreta no está definida para los valores intermedios de tiempo.</li>
</ul>




```python
import numpy as np
import matplotlib.pyplot as plt

# ===========================
# Parámetros de la señal
# ===========================
f = 2  # frecuencia de la señal (Hz)
t_cont = np.linspace(0, 1, 1000)  # tiempo continuo
x_cont = np.sin(2 * np.pi * f * t_cont)  # señal continua

# ===========================
# Muestreo
# ===========================
fs = 50  # frecuencia de muestreo (Hz) -> alta para apreciar bien el muestreo
t_samp = np.arange(0, 1, 1/fs)
x_samp = np.sin(2 * np.pi * f * t_samp)

# ===========================
# Gráficas
# ===========================
plt.figure(figsize=(9, 6))

# Señal continua
plt.subplot(2, 1, 1)
plt.plot(t_cont, x_cont, 'b', linewidth=2)
plt.title('Señal continua')
plt.ylabel('Amplitud')
plt.xlim(0, 1)        # Alinear eje x
plt.grid(True)

# Señal muestreada (pulsos)
plt.subplot(2, 1, 2)
plt.stem(t_samp, x_samp, linefmt='r-', markerfmt='ro', basefmt='k')
plt.title('Señal muestreada')
plt.xlabel('Tiempo [s]')
plt.ylabel('Amplitud')
plt.xlim(0, 1)        # Alinear eje x
plt.grid(True)

plt.tight_layout()
plt.show()
```

<h1 style="font-size:200%; text-align:left;">
Existen múltiples dispositivos de adquisición de imágenes 2D, entre ellos podemos citar:
</h1>

<ul style="font-size:170%; line-height:35px;">
  <li><span style="color:blue">Cámara CCD:</span> Charge-Coupled Device (CCD) es el más popular de los dispositivos de adquisición de imágenes bi-dimensionales.</li>
  <li><span style="color:blue">Infrarrojo:</span> Utilizadas en termografía, estas cámaras permiten obtener imágenes en función de la temperatura en la escena.</li>

  <li><span style="color:blue">Imágenes hiperespectrales:</span> Capturan información en <strong>múltiples longitudes de onda (visible, infrarrojo cercano, etc.)</strong>, lo cual permite analizar propiedades físicas o químicas de los objetos. Son comunes en teledetección, agricultura de precisión y análisis ambiental.</li>
  <li><span style="color:blue">Ultrasonido:</span> Utilizadas en medicina (reflexión de ondas sonoras).</li>
</ul>

  <li>$D(p, z) \le D(p, q) + D(q, z) \quad \rightarrow$ <b>desigualdad triangular</b>.</li>

<h1 style="font-size:200%; color:blue; margin-top:25px;">
Ecualización del Histograma de una imagen
</h1>

<ul style="font-size:170%; line-height:35px; margin-top:20px;">
  <li>Es una transformación que busca obtener, para una imagen, un <span style="color:blue;">histograma con una distribución uniforme</span>.</li>
  <li>Es decir, que exista el mismo número de píxeles para cada nivel de gris del histograma.</li>
  <li>Está definida como:</li>
</ul>

<p style="font-size:170%; text-align:center; line-height:35px;">
$$
h(v) = \text{round} \left( \frac{cdf(v) - cdf_{min}}{(M \times N) - 1} \times (L-1) \right)
$$
</p>

<ul style="font-size:170%; line-height:35px;">
  <li>donde:</li>
  <ul>
    <li>$v$ es la intensidad de gris del píxel.</li>
    <li>$cdf$ es la función de distribución acumulativa.</li>
    <li>$cdf_{min}$ es el mínimo valor, diferente de cero, de la función de distribución acumulativa.</li>
    <li>$M \times N$ es el número de píxeles en la imagen.</li>
    <li>$L$ es el número de niveles de gris.</li>
  </ul>
</ul>


```python
import cv2
import matplotlib.pyplot as plt

# Ecualización del histograma
equ = cv2.equalizeHist(img)

# Crear figura más grande
plt.figure(figsize=(12, 10))

# Imagen original
plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original', fontsize=14)
plt.axis('off')

# Imagen ecualizada
plt.subplot(222)
plt.imshow(equ, cmap='gray')
plt.title('Imagen Ecualizada', fontsize=14)
plt.axis('off')

# Histograma de la imagen original
plt.subplot(223)
plt.hist(img.ravel(), bins=128, range=[0,256], color='b')
plt.title('Histograma Original', fontsize=14)
plt.xlabel('Nivel de gris')
plt.ylabel('Frecuencia')

# Histograma de la imagen ecualizada
plt.subplot(224)
plt.hist(equ.ravel(), bins=128, range=[0,256], color='r')
plt.title('Histograma Ecualizado', fontsize=14)
plt.xlabel('Nivel de gris')
plt.ylabel('Frecuencia')

# Ajuste automático de espacios
plt.tight_layout()
plt.show()
```

<h4 style="text-align:center;">Imagen tomada de: <a href="http://web.pdx.edu/~jduh/courses/Archive/geog481w07/Students/Ludwig_ImageConvolution.pdf" target="_blank">fuente</a></h4>


```python
import numpy as np

# Definición del kernel y la señal
kernel = np.array([[7, 8, 3], [1, 1, 0], [1, 2, 1]], dtype=int)
signal = np.array([[15, 20, 101, 100],
                   [200, 50, 55, 8],
                   [10, 11, 230, 202],
                   [100, 130, 115, 120]], dtype=int)

print("Kernel:\n", kernel)
print("\nSeñal:\n", signal)

# Tamaño de la salida
rows = signal.shape[0] - kernel.shape[0] + 1
cols = signal.shape[1] - kernel.shape[1] + 1
output = np.zeros((rows, cols), dtype=int)
print("\n(Rows: %d, Cols: %d)" % (rows, cols))

# Kernel invertido (rotado 180°)
kernel_reversed = np.rot90(kernel, 2)
print("\nKernel invertido:\n", kernel_reversed)

# Convolución manual
for i in range(rows):
    for j in range(cols):
        patch = signal[i:i+kernel.shape[0], j:j+kernel.shape[1]]
        output[i, j] = np.sum(patch * kernel_reversed)

print("\nSalida de la convolución:\n", output)
```

## Convolución

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Definir kernel
kernel1 = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]], dtype=np.int16)

# Cargar imagen en escala de grises
img = cv2.imread('fig/Notebook/Logo.png', 0)

# Tamaño de la salida
rows = img.shape[0] - kernel1.shape[0] + 1
cols = img.shape[1] - kernel1.shape[1] + 1
img_salida = np.zeros((rows, cols), dtype=np.int16)
img_salida_abs = np.zeros((rows, cols), dtype=np.int16)

# Rotación 180° del kernel
kernel_reversed = np.rot90(kernel1, 2)

# Convolución manual
for i in range(rows):
    for j in range(cols):
        patch = img[i:i+kernel1.shape[0], j:j+kernel1.shape[1]]
        conv_value = np.sum(kernel_reversed * patch)
        img_salida[i, j] = conv_value
        img_salida_abs[i, j] = abs(conv_value)

# Visualización con figura más grande
plt.figure(figsize=(12, 8))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original', fontsize=14)
plt.axis('off')

plt.subplot(222)
plt.imshow(img_salida, cmap='gray')
plt.title('Convolución (sin abs)', fontsize=14)
plt.axis('off')

plt.subplot(224)
plt.imshow(img_salida_abs, cmap='gray')
plt.title('Convolución (con abs)', fontsize=14)
plt.axis('off')

# Ajuste automático de espacios
plt.tight_layout()
plt.show()
```

## Mediana

```python
import numpy as np

# Definir kernel y señal
kernel = np.array([[7, 8, 3], [1, 1, 0], [1, 2, 1]], dtype=np.uint8)
Imagen = np.array([[15, 20, 101, 100],
                   [200, 50, 55, 8],
                   [10, 11, 230, 202],
                   [100, 130, 115, 120]], dtype=np.uint8)

print("Kernel:\n", kernel)
print("\nImagen:\n", Imagen)

# Tamaño de la salida
rows = Imagen.shape[0] - kernel.shape[0] + 1
cols = Imagen.shape[1] - kernel.shape[1] + 1
output = np.zeros((rows, cols), dtype=np.uint8)
print("\nSalida: Filas = %d, Columnas = %d" % (rows, cols))

# Kernel invertido (aunque para mediana no es estrictamente necesario)
kernel_reversed = np.rot90(kernel, 2)

# Filtro mediana manual
for i in range(rows):
    for j in range(cols):
        patch = Imagen[i:i+kernel.shape[0], j:j+kernel.shape[1]]
        patch_flat = patch.flatten()
        patch_sorted = np.sort(patch_flat)
        output[i, j] = patch_sorted[len(patch_sorted)//2]  # mediana

print("\nÚltimo patch procesado:\n", patch)
print("\nVector patch sin ordenar:", patch_flat)
print("Vector patch ordenado:", patch_sorted)
print("Valor de la mediana:", patch_sorted[len(patch_sorted)//2])

print("\nSalida final:\n", output)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
M = 3  # Tamaño del filtro mediana
filas, columnas = 500, 500

# Crear imagen original
img = np.zeros((filas, columnas), np.uint8)
img[filas//4:3*filas//4, columnas//4:3*columnas//4] = 255

# Agregar ruido sal y pimienta (10%)
ruido1 = img.copy()
probabilidad = 0.10
aleatorio = np.random.rand(filas, columnas)
ruido1[(aleatorio < probabilidad/2) & (img == 0)] = 255  # Sal
ruido1[(aleatorio < probabilidad/2) & (img == 255)] = 0   # Pimienta

# Filtrado mediana
pad = M // 2
output = np.zeros_like(ruido1)
for i in range(pad, filas-pad):
    for j in range(pad, columnas-pad):
        patch = ruido1[i-pad:i+pad+1, j-pad:j+pad+1].flatten()
        output[i, j] = np.median(patch).astype(np.uint8)

# Visualización
plt.figure(figsize=(12, 10))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original', fontsize=14)
plt.axis('off')

plt.subplot(222)
plt.imshow(ruido1, cmap='gray')
plt.title('Imagen con Ruido 10%', fontsize=14)
plt.axis('off')

plt.subplot(224)
plt.imshow(output, cmap='gray')
plt.title('Filtro Mediana', fontsize=14)
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from imageio import imread

# -------------------------------
# Cargar la imagen
# -------------------------------
img = imread('fig/Notebook/Imagen_bitmap_1.jpg', pilmode='L')  # cargar como gris
N, M = img.shape

# -------------------------------
# Transformada de Fourier
# -------------------------------
F = np.fft.fft2(img)                # TF
F_shift = np.fft.fftshift(F)        # TF centrada
F_mag = np.abs(F)                   # Magnitud
F_shift_mag = np.abs(F_shift)       # Magnitud centrada

# Normalización para mostrar
F_shift_mag_norm = F_shift_mag / np.max(F_shift_mag)

# -------------------------------
# Mostrar resultados
# -------------------------------
plt.figure(figsize=(12,10))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(np.log1p(F_mag), cmap='gray')  # log para visualizar mejor
plt.title('TF (Magnitude)')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(np.log1p(F_shift_mag), cmap='gray')
plt.title('TF Centrada (Magnitude)')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(np.log1p(F_shift_mag_norm), cmap='gray')
plt.title('TF Centrada y Normalizada')
plt.axis('off')

plt.tight_layout()
plt.show()
```

<h1 style="font-size:200%; text-align:center;">
<span style="color:blue">Operaciones geométricas</span>
</h1>

<ul style="font-size:170%; line-height:35px;">
  <li>La posición de cada píxel en la imagen resultado depende de la posición en la imagen origen.</li>
  <ul>
      <li>Zoom</li>
      <li>Traslación</li>
      <li>Rotación</li>
      <li>Warping</li>
  </ul>
  <li>En general las operaciones geométricas requieren de <b>interpolación</b> o <b>submuestreo</b> de píxeles.</li>
</ul>


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------
# Cargar imagen
# -------------------------------
img = cv2.imread('fig/Notebook/Imagen_bitmap_1.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # convertir a RGB
h, w = img.shape[:2]

# -------------------------------
# 1. Zoom (Escalado)
# -------------------------------
zoom_factor = 1.5
zoomed = cv2.resize(img, None, fx=zoom_factor, fy=zoom_factor, interpolation=cv2.INTER_LINEAR)

# -------------------------------
# 2. Traslación
# -------------------------------
tx, ty = 50, 30  # traslación en x e y
M_translate = np.float32([[1, 0, tx],
                          [0, 1, ty]])
translated = cv2.warpAffine(img, M_translate, (w, h))

# -------------------------------
# 3. Rotación
# -------------------------------
angle = 45  # grados
center = (w//2, h//2)
scale = 1.0
M_rotate = cv2.getRotationMatrix2D(center, angle, scale)
rotated = cv2.warpAffine(img, M_rotate, (w, h))

# -------------------------------
# 4. Warping (Transformación afín)
# -------------------------------
pts1 = np.float32([[0,0], [w-1,0], [0,h-1]])
pts2 = np.float32([[0,0], [w*0.8, h*0.2], [w*0.2, h*0.8]])
M_affine = cv2.getAffineTransform(pts1, pts2)
warped = cv2.warpAffine(img, M_affine, (w, h))

# -------------------------------
# Mostrar resultados
# -------------------------------
plt.figure(figsize=(12,10))

plt.subplot(2,2,1)
plt.imshow(zoomed)
plt.title('Zoom')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(translated)
plt.title('Traslación')
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(rotated)
plt.title('Rotación')
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(warped)
plt.title('Warping')
plt.axis('off')

plt.tight_layout()
plt.show()
```
